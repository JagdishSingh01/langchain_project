### Models Integration with google gemini and groq

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GEMINI_API_KEY"]=os.getenv("GEMINI_API_KEY_1")
os.environ["GROQ_API_KEY"]= os.getenv("GROQ_API_KEY")
```

### GOOGLE GEMINI MODEL INTEGRATION

```python
import os
from langchain.chat_models import init_chat_model
model= init_chat_model("google_genai:gemini-3.1-flash-lite")
response= model.invoke("why do we have seasons on earth?")
response
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
response = model.invoke("why do parrots talk?")
response
```

### GROQ MODEL INTEGRATION

```python
model= init_chat_model("groq:openai/gpt-oss-120b")
response= model.invoke("why do we have seasons on earth?")
response
```

```python
from langchain_groq import ChatGroq
model = ChatGroq(model="openai/gpt-oss-120b")
response = model.invoke("who was the missile man?")
response
```

```python
for chunk in model.stream("Why do parrots have colorful feathers?"):
  print(chunk.text, end="", flush=True)
```

```python
responses = model.batch([
  "why do parrots have colorful feathers?",
  "why do we have seasons on earth?",
  "how do airplanes fly?",
])
for response in responses:
  print(response)
```

```python
responses = model.batch(
  [
    "why do parrots have colorful feathers?",
    "why do we have seasons on earth?",
    "how do airplanes fly?",
    "who was the missile man?",
    "how hot air balloons work?",
    "how is pencil made?",
  ],
  config={"max_concurrency": 5}
)
for response in responses:
  print(response)
```
